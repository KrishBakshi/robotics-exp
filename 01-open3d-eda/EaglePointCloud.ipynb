```python
import numpy as np
import open3d as o3d
import plotly.graph_objects as go

pcd = o3d.io.read_point_cloud(o3d.data.EaglePointCloud().path)
pts = np.asarray(pcd.points)
cols = np.asarray(pcd.colors) if pcd.has_colors() else None

# downsample so the browser stays responsive
idx = np.random.default_rng(0).choice(len(pts), size=min(80_000, len(pts)), replace=False)
pts, cols = pts[idx], (None if cols is None else cols[idx])

fig = go.Figure(data=[go.Scatter3d(
    x=pts[:, 0], y=pts[:, 1], z=pts[:, 2],
    mode="markers",
    marker=dict(size=1, color=cols if cols is not None else pts[:, 2]),
)])
fig.update_layout(width=800, height=600, scene_aspectmode="data")
fig.show()
```
